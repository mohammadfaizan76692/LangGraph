```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Annotated
from langchain_core.messages import BaseMessage, HumanMessage, AIMessage, SystemMessage
from langchain_openai import ChatOpenAI
```

```python
## reducer  function helps to append output in state ,if same node runs more than one time
from langgraph.graph.message import add_messages

## create state every node will read from state or update this state
class ChatState(TypedDict):

    messages: Annotated[list[BaseMessage], add_messages]
```

```python
llm = ChatOpenAI()

def chatnode(state: ChatState):

    messages = state['messages']

    response = llm.invoke(messages)

    return {'messages':[response]}



```

```python
graph = StateGraph(ChatState)
```

```python
graph.add_node('chat_node', chatnode)
graph.add_edge(START,'chat_node')
graph.add_edge('chat_node', END)
```

```python
chatbot = graph.compile()
```

```python
initial_state = {"messages":[]}

while True:
    user_input = input()

    if user_input.strip().lower()=='exit':
        break

    print(f"Human: {user_input}")
    initial_state['messages'].append(HumanMessage(content=user_input))
    response = chatbot.invoke(initial_state)['messages'][-1].content

    print(f"AI: {response}")
```
